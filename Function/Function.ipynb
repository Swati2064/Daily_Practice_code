```python
def add_sub(x,y):
    c=x+y
    d=x-y
    return c,d

r,r1= add_sub(4,5)
print(r)
print(r1)

print(type(r))
print(type(r1))
```

```python
def add_sub_mul(x,y):
    c=x+y
    d=x-y
    e=x*y
    return c,d,e

r,r1,r2=add_sub_mul(4,5)
print(r)
print(r1)
print(r2)

print(type(r))
print(type(r1))
```

# positional argument

```python
def person(name,age):
    print(name)
    print(age)

person('nit',23)
```

# keyword argument

```python
def person(name,age):
    print(name)
    print(age+1)

person(age=23,name='nit')
```

```python
def person(name,age,age2):
    print(name)
    print(age)
    print(age2)

person(age=20,name='nit',age2=30)
```
