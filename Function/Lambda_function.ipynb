# Lambda Function 

```python
# Lambda function is use to reduce lines
# one argument
square=lambda x: x*x
print(square(3))
```

```python
# use condition 
age=lambda age:("young") if age < 45 else ("old")
print(age(44))
```

```python
# use condition 
age=lambda age:("young") if age < 45 else ("old")
print(age(50))
```

```python
# 2 argumnets
num=lambda a,b:("smallest") if a > b else ("largest")
print(num(20,10))
```

```python
num=lambda a,b: ("smallest") if a > b else ("largest")
print(num(10,60))
```

```python
# 3 arguments
num=lambda a,b,c:("smallest") if a>b>c else("largest")
print(num(10,20,30))
```

```python
num=lambda a,b,c:("smallest")if a>b>c else("largest")
print(num(50,30,20))
```

```python
# Zip use for combine
name=["smith","alex"]
age=[20,30]
print(list(zip(name,age)))
```

```python
name=["xyz","abc"]
age=[10,20]
print(list(zip(name,age)))
```

```python
# no default function
greet=lambda: "hello"
print(greet())
```

```python
# even or odd function
even_odd=lambda num:"even" if num%2==0 else "odd"
print(even_odd(22))
```

```python
even_odd=lambda num:"even" if num%2==0 else "odd"
print(even_odd(25))
```

```python
# Lambda with map function use same like for loop
num=[1,2,5,4,3,7]
result=list(map(lambda x:x**2,num))
print(result)
```

```python
num=[1,2,3,4,5]
result=list(map(lambda x:x**3,num))
print(result)
```

```python
# lambda with filter()
nums=[5,6,8,2,9]
result=list(filter(lambda x:x%2==0,nums))
print(result)
```

```python
# lambda with filter()
nums=[3,1,5,6,9,6,8,2,4]
result=list(filter(lambda x:x%3==0,nums))
print(result)
```

```python
# Lamda with sorted()
# sort by length
names=["smith","john","alex","roy","tom"]
result=sorted(names,key=lambda x:len(x))
print(result)
```

```python
# lambda with max()
nums=[10,20,50,60,90]
largest=max(nums,key=lambda x:x)
print(largest)
```

```python
# Lambda with min()
nums=[50,30,60,80,10]
smallest=min(nums,key=lambda x:x)
print(smallest)
```

```python
# lambda with reduce()
from functools import reduce
result=reduce(lambda x,y:x+y,[1,2,3,4,5])
print(result)
```

```python
# lambda with sort()
# sort Tuples
students=[("alex",20),("amit",23),("roy",25)]
students.sort(key=lambda x:x[1])
print(students)
```

```python
# lambda with Dictinoary
students={"john":52,"jay":80,"reem":25}
result=sorted(students.items(),key=lambda x:x[1])
print(result)
```

```python
# Lambda with multiple conditions
check=lambda x:("Positive Even" if x > 0 and x % 2 == 0 else "Other")
print(check(8))
```

```python
# find largest num
largest=lambda a,b: a if a > b else b
print(largest(10,30))
```

```python
# Find the String
result=lambda s:len(s)
print(result("Python"))
```
