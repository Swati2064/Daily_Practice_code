```python
# Argument vs Parameter
def add(a,b):
    print(a+b) # a and b → parameters(Parameter = variable in function definition)
add(5,9) # 10 and 20 → arguments(Argument = actual value passed during function call)
```

```python
# return Statement
# return sends a value back from the function.
def add(a,b):
    return a+b # return sends the result back so it can be stored or used later.
result=add(10,20)
print(result)
```

```python
# Function with Multiple Parameters
def student(name,age,marks):
    print(name)
    print(age)
    print(marks)
student("Swati",22,86)
```

```python
# Positional Arguments
# Values are assigned according to their position.
def student(name,age):
    print(name,age)
student("swati",22)
```

```python
# Keyword Arguments
# We specify the parameter name.

def student(name,age):
    print(name,age)
student(age=22,name="swati")
```

```python
# Default Arguments
# We can provide a default value.
def greet(name="student"):
    print("hello",name)
greet()
greet("swati")
```

```python
# *args
# *args allows a function to accept multiple positional arguments.

def add(*numbers):
    total=0
    for num in numbers:
        total+=num
    return total

print(add(10,20))
print(add(60,30,40,66))
```

```python
# *args
# *args allows a function to accept multiple positional arguments.
def add(*numbers):
    total=0
    for num in numbers:
        total+=num
    return total
print(add(80,20))
print(add(20,10,50,30))
```

```python
# **kwargs
# **kwargs allows multiple keyword arguments.
def student(**details):
    print(details)
student(name="swati",age=22,city="Dharashiv")
```

```python
# Local Variable
# A variable created inside a function is generally a local variable.
def test():
    x=10
    print(x)
test()
```

```python
# Global Variable
# A variable created outside a function is a global variable

x=100
def test():
    print(x)
test()
```

```python
# Important Function Programs
# Program 1 — Add Two Numbers

def add(a,b):
    return a+b
print(add(10,20))
```

```python
# Program 2 — Check Even or Odd
def check_even_odd(num):
    if num % 2 ==0:
        return "Even"
    else:
        return "Odd"
print(check_even_odd(10))
```

```python
# Program 3 — Find Maximum
def find_max(a,b):
    if a>b:
        return a
    else:
        return b
print(find_max(10,20))
```

```python
# Program 4 — Factorial Using Function
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
print(factorial(5))
```
