```python
# elif means else if. It is used to check another condition when the previous condition is False.

temperature=40

if temperature >= 35:
    print("very hot")
elif temperature >= 25:
    print("warm")
elif temperature >= 15:
    print("cool")
else:
    print("cold")
```
