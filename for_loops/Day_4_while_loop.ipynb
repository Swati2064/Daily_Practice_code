# While loop

```python
# A while loop is used to repeatedly execute a block of code as long as a given condition is True.
i=1

while i <= 5:
    print(i)
    i+=1
```

```python
# while Loop with break

# break is used to immediately stop the loop.
i=1

while i <= 10:
    if i == 5:
        break
    print(i)
    i+=1
```

```python
# while Loop with continue

# continue skips the current iteration.

i=0
while i<5:
    i+=1
    if i==3:
        continue
    print(i)
```

```python
# Check palindrome number
num=121
original=num
reverse=0

while num>0:
    digit=num%10
    reverse=reverse*10+digit
    num//=10
if original==reverse:
    print("palindrome")
else:
    print("not palindrome")
    
```
