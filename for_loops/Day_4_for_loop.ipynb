```python
# A loop is used to execute the same block of code multiple times.

for i in range(5):
    print("hello")
```

```python
# A for loop is generally used when you want to iterate over a sequence or 
# when you know how many times you want to repeat something.
# syntax - for variable in sequence:
# range() is commonly used with for loops.

for i in range(7): # range(stop)
    print(i)
```

```python
for i in range(1,7): # range(start, stop)
    print(i)
```

```python
for i in range(1,15,3): # range(start, stop, step)
    print(i)
```

```python
# Loop Through a String

name="ram"

for ch in name:
    print(ch)
```

```python
# Loop Through a List

numbers=[10,20,30,50,40]

for num in numbers:
    print(num)
```

```python
# continue skips the current iteration and moves to the next iteration.

for i in range(1,9):
    if i == 3:
        continue
    print(i)
```

```python
# A loop inside another loop is called a nested loop.

for i in range(1,4):
    for j in range(1,4):
        print(i,j)
```
