```python
# Basic for loop
names=["sumit","ravi","aman"]

for name in names:
    print(name)
```

```python
# for loop with range()

for i in range(5):
    print(i)
```

```python
# start and stop

for i in range(2,6):
    print(i)
```

```python
# start, stop, step

for i in range(1,10,2):
    print(i)
```

```python
# Reverse loop

for i in range(10,0,-1):
    print(i)
```

```python
# Loop Through string

word="python"

for ch in word:
    print(ch)
```

```python
# loop through list

nums=[10,20,30]

for i in nums:
    print(i)
```

```python
# loop through tuple

data=(1,2,3)

for x in data:
    print(x)
```

```python
# loop through dictionary
# keys
student={"name":"sumit","age":22}

for key in student:
    print(key)
```

```python
# values
for value in student.values():
    print(value)
    
```

```python
# key and values both

for key,value in student.items():
    print(key,value)
```

```python
# Using enumerate()
# used for index + values together

names=["A","B","C"]

for index,value in enumerate(names):
    print(index,value)
```

```python
# Nested for Loop

for i in range(3):
    for j in range(2):
        print(i,j)
```

```python
# break Statement
# stop loop immediately.

for i in range(10):
    if i==6:
        break
    print(i)
```

```python
# continue statement
# skips current iteration.

for i in range(5):
    if i==2:
        continue
    print(i)
```

```python
# pass statement
# placehoder

for i in range(5):
    pass
```

```python
# for-else loop
# else runs if loop finishes normally.

for i in range(5):
    print(i)
else:
    print("Loop completeted")
```

```python
# loop through 2 lists using zip()
names=["A","B","C"]
marks=[80,90,70]

for n,m in zip(names,marks):
    print(n,m)
```

```python
# list comprehension (for in single line)

square=[x*x for x in range(5)]

print(square)
```

```python
# loop through set

nums={1,2,3}

for n in nums:
    print(n)
```

```python
# iterate 2D list

matrix=[[1,2],[3,4]]

for row in matrix:
    for value in row:
        print(value)
```

```python
# Unpacking in for loop

data=[(1,2),(3,4),(5,6)]

for a,b in data:
    print(a,b)
```
